# The command line

Work through this notebook from top to bottom. This week you type most
commands in a **terminal**, not in a notebook cell. Keep this notebook and a
terminal open side by side.

The *command line* (also called the *shell* or the *terminal*) is a way to
use a computer by typing commands instead of clicking. You type one command,
press Enter, and the computer answers with text.

## Why would a biologist use the command line?

Most of you have never needed it. Here is when you will:

- **The tools are there.** Much of bioinformatics has no buttons: BLAST,
  sequence aligners, tools for quality control of sequencing data, protein
  structure prediction. The methods section of a paper says "we ran X with
  options Y". To reuse that method, you run the same command.
- **The data are too big for your laptop.** A sequencing run, a long
  microscopy movie or a week of animal tracking data lives on a server or a
  computing cluster. These computers have no desktop and no mouse. You log in
  and type.
- **Doing the same thing 1000 times.** Renaming 2000 microscope images, or
  running the same analysis on every file of an experiment overnight.
  Clicking does not scale. A command with a loop or a wildcard does.
- **A written record.** A command is exact. You can paste it into your lab
  notebook or your methods section, and someone else can repeat it. A series
  of clicks is hard to write down and hard to repeat.
- **Lab hardware.** Many lab setups (cameras, behaviour rigs, small computers
  like a Raspberry Pi) run Linux without a screen. You control them from the
  command line.
- **AI coding assistants** (later in this course) increasingly work in the
  terminal and run commands for you. You need to be able to read what they do
  before you allow it.

When *not* to use it: for one small table or a one-off plot, a spreadsheet
or a notebook is fine. The command line is a tool for scale, for repetition
and for working on other computers.

## The prompt

The text before the cursor, ending in `$`, is the **prompt**. It means "I am
ready, type a command". In this notebook we write commands like this:

```
$ pwd
```

Do **not** type the `$`. Type only `pwd` and press Enter.

Many commands print nothing when they work. No news is good news.

## Where am I? `pwd`

Your files are organised in **directories** (another word for folders). At
any moment the terminal is "in" one directory: the **working directory**. It
is like the folder that is open in a file browser.

`pwd` ("print working directory") prints it:

```
$ pwd
/home/yourname
```

When you open a new terminal, you are in your **home directory**. (Your name
and the exact path may look different.)

## What is here? `ls`

`ls` ("list") shows what is in the working directory:

```
$ ls
```

Commands can take **options**, which start with `-`. They change what the
command does:

- `ls -l` ("long") shows one line per file, with its size and date.
- `ls -a` ("all") also shows **hidden** files and directories. Their names
  start with a dot, like `.config`. They are hidden only to keep the normal
  list short.
- Options can be combined: `ls -la`.

You can also give `ls` the name of a directory to look into:

```
$ ls /schnitzeljagd
```

Try all of these now.

## Moving around: `cd`

`cd` ("change directory") moves you into another directory:

```
$ cd /schnitzeljagd
$ pwd
/schnitzeljagd
```

Two special names:

- `..` means "the directory above this one" (the *parent*).
- `.` means "this directory".

So `cd ..` moves one level up. `cd` on its own takes you back home.

Try it: go to `/schnitzeljagd`, look around with `ls`, go up with `cd ..`,
and go home with `cd`. Check where you are with `pwd` after each step.

## Paths

A **path** says where a file or directory is. Directory names are separated
by `/`. Imagine this part of a computer:

```
/
├── home/
│   └── anna/
│       ├── PM-21/
│       │   └── notes.txt
│       └── photos/
└── schnitzeljagd/
```

- An **absolute path** starts with `/`, the very top (the *root*). It works
  from anywhere: `/home/anna/PM-21/notes.txt`.
- A **relative path** does not start with `/`. It starts from the working
  directory. If Anna is in `/home/anna`, then `PM-21/notes.txt` means the same
  file.
- `..` works inside paths, too. From `/home/anna/PM-21`, the path
  `../photos` means `/home/anna/photos`.
- `~` is short for your home directory: `~/PM-21` is `/home/anna/PM-21` for
  Anna.

Avoid spaces in file and directory names. On the command line a space
separates one thing from the next, so `cd PM 21` means "cd with two
arguments, `PM` and `21`". Use `PM-21` or `PM_21` instead.

## Reading a file: `cat` and `less`

`cat` prints a whole file:

```
$ cat /schnitzeljagd/README.md
```

That is the start of this week's exercise. (`.md` files are *Markdown*: plain
text, the same format as the text cells in a notebook. A `README.md` file
usually explains what is in a directory.)

For a long file, `less` shows one page at a time. Use the arrow keys or the
space bar to move, and press `q` to quit.

## Three keys that save a lot of typing

- **Tab** completes names. Type `cd /schn` and press Tab: the terminal fills
  in `/schnitzeljagd/`. If nothing happens, press Tab twice to see the
  choices. Use Tab all the time: it is faster and it avoids typos.
- **Up arrow** brings back the previous command. Press it again for older
  ones. Edit the command with the left and right arrows, then press Enter.
- **Ctrl-C** stops a command that is running (or that is waiting for you).
  In a terminal, Ctrl-C does *not* copy text. To copy and paste on macOS,
  use Cmd-C and Cmd-V as usual. On Windows and Linux, try Ctrl-Shift-C and
  Ctrl-Shift-V, or the right mouse button.

The mouse cannot move the cursor in a terminal. Use the arrow keys.

## When something goes wrong

- `command not found`: a typo in the command name, or the program is not
  installed.
- `No such file or directory`: the path is wrong. Check where you are with
  `pwd` and what is there with `ls`. Use Tab to complete names.
- The terminal seems stuck, or shows no prompt: press Ctrl-C. In `less`,
  press `q`.

Error messages in the terminal are like error messages in Python: read them,
they usually say what is wrong.

## Summary

| Command | What it does |
| ------- | ------------ |
| `pwd` | print the working directory |
| `ls` | list files (`-l` long, `-a` also hidden) |
| `cd dir` | change into `dir` (`cd ..` up, `cd` home) |
| `cat file` | print a file |
| `less file` | view a file page by page (`q` quits) |

## Going further (optional)

- `ls -lh` shows file sizes in a human-readable form (`K`, `M`, `G`).
- Most commands explain themselves: `ls --help`, or `man ls` (the manual;
  `q` quits).
- `history` lists your previous commands. Ctrl-R searches them: press Ctrl-R
  and type part of an old command.
- `clear` (or Ctrl-L) clears the screen.
- `cd -` jumps back to the directory you were in before.
