## Edit a text file: `nano`

`nano` is a simple text editor that runs inside the terminal:

```
$ nano sample1.txt
```

Type a DNA sequence, for example `ATGCGCGATTACA`. Then:

- **Ctrl-O** writes ("saves") the file. Press Enter to confirm the name.
- **Ctrl-X** exits nano.

The bottom two lines of nano list the commands. `^` means Ctrl, so `^X` is
Ctrl-X. The mouse does not work in nano; use the arrow keys.

Check the result with `cat sample1.txt`.

This is a **plain text** file: just characters, no fonts or colours. Many file
types are plain text: `.txt`, `.csv`, `.md` and Python scripts, `.py`. A Word
document (`.docx`) is not; do not use Word to edit code or data files.

## Quick files with `echo` and `>`

`echo` prints its arguments:

```
$ echo hello
hello
```

With `>`, the output goes into a file instead of the screen:

```
$ echo GGGCCCATAT > sample2.txt
$ cat sample2.txt
GGGCCCATAT
```

Careful: `>` **replaces** the file if it already exists. `>>` adds to the end
of the file instead.

```
$ echo ATATATGC >> sample2.txt
$ cat sample2.txt
GGGCCCATAT
ATATATGC
```

## Copy, move, rename: `cp` and `mv`

```
$ cp sample1.txt sample1-backup.txt
$ ls
```

`cp source destination` copies. If the destination is a directory, the copy
goes into it, with the same name:

```
$ mkdir backup
$ cp sample2.txt backup
$ ls backup
```

`mv` moves a file. Moving within the same directory is how you **rename**:

```
$ mv sample1-backup.txt old-sample1.txt
$ ls
```

## Delete: `rm`

```
$ rm old-sample1.txt
```

**There is no recycle bin.** `rm` deletes immediately, with no "are you
sure?" and no undo. Check the name twice before pressing Enter.

`rm` deletes files. For an empty directory use `rmdir`. `rm -r` deletes a
directory with everything in it; use it rarely, and carefully.

## Many files at once: the wildcard `*`

`*` stands for "any characters". The shell replaces the pattern by all
matching names *before* the command runs:

```
$ ls *.txt
$ cat sample*.txt
$ cp *.txt backup
```

This is where the command line becomes powerful. `cp *.txt backup` is the
same amount of typing for 2 files or for 2000.

## Looking into files: `head`, `tail`, `wc`

Data files can be huge. Before you open a big file, look at a small piece:

- `head file` prints the first 10 lines, `head -n 3 file` the first 3.
- `tail file` prints the last 10 lines.
- `wc -l file` counts the lines. For a CSV file with one row per
  measurement, that is (almost) the number of measurements.

Try them on `sample2.txt` and on `/schnitzeljagd/README.md`.

## Summary

| Command | What it does |
| ------- | ------------ |
| `mkdir dir` | make a directory |
| `nano file` | edit a text file (Ctrl-O save, Ctrl-X exit) |
| `echo text > file` | write text into a file (`>>` adds to the end) |
| `cp a b` | copy |
| `mv a b` | move or rename |
| `rm file` | delete, with no undo |
| `*` | wildcard: any characters |
| `head`, `tail`, `wc -l` | start, end, number of lines of a file |

## Going further (optional)

- **Pipes.** `|` sends the output of one command into the next one.
  `ls | wc -l` counts the files in a directory.
- **Searching.** `grep ATG sample2.txt` prints the lines that contain `ATG`.
  `grep -r word directory` searches all files in a directory.
- **Safer deleting.** `rm -i file` asks before deleting.
- `ls -lh` and `du -sh directory` show how much space files take.
