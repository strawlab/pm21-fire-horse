## Working in Jupyter Lab

A notebook is a list of **cells**. There are two kinds you need now:

- **Code cells** contain Python. When you run one, the result appears below it.
  The number to the left, like `[3]`, shows the order in which cells were run.
- **Markdown cells** contain text, like this one. Markdown is plain text with a
  few formatting rules: `**bold**` gives **bold**, `*italic*` gives *italic*,
  a line starting with `#` is a heading, and text in backticks like `` `x` ``
  is shown as code.

### Running and editing cells

- **Run** the selected cell: press **Shift-Enter**. This also moves to the
  next cell.
- **Edit** a markdown cell: **double-click** it. Press Shift-Enter to show it
  formatted again.
- **Add** a cell: click the `+` button in the toolbar.
- **Change** a cell's type: use the drop-down menu in the toolbar ("Code" or
  "Markdown").

Try it: double-click this cell to see its markdown text, then press
Shift-Enter.

### Order matters

Python remembers what you ran, not what you see on the screen. If you change a
cell, you must run it again. If you run cells out of order, the results can be
confusing. When that happens, use **Kernel → Restart Kernel and Run All
Cells...** to start fresh and run everything from top to bottom.

### Saving

Jupyter saves automatically every few minutes, but save before you close a
notebook or upload it: **File → Save Notebook** (or Ctrl-S, on macOS Cmd-S).
