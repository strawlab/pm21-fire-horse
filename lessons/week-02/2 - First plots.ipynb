## Importing matplotlib

Python has only a few built-in functions. Many more tools come in
**packages** (also called *libraries*): collections of Python code written by
other people. You load a package with `import`. The most common package for
plots and figures is *matplotlib*. It is not part of Python itself. Sunbeam
installed it for you together with Jupyter Lab.

The line below loads the `pyplot` part of matplotlib and gives it the short
name `plt`. Many packages have a standard short name like this, and you will
see exactly this line in a huge amount of Python code. For now, just copy it;
we come back to `import` later.

````python
import matplotlib.pyplot as plt
````

## A line plot

`plt.plot(x, y)` draws a line through points. The list `x` gives the
horizontal positions and the list `y` the vertical positions.

Here is the optical density (OD600) of a bacterial culture, measured every
hour. The OD rises as the bacteria grow.

````python
hours = [0, 1, 2, 3, 4, 5, 6]
od600 = [0.05, 0.08, 0.15, 0.30, 0.55, 0.80, 0.95]
plt.plot(hours, od600)
````

Above the plot you may see a line of text such as
`[<matplotlib.lines.Line2D at 0x...>]`. This is the return value of the last
line of the cell, which Jupyter always shows. You can ignore it. In the cells
below you will see similar lines, such as `Text(...)`.

## Labels

A plot without axis labels is hard to read. Add them with `plt.xlabel()`,
`plt.ylabel()` and `plt.title()`. All calls in the same cell draw into the
same figure.

````python
plt.plot(hours, od600)
plt.xlabel("time (h)")
plt.ylabel("OD600")
plt.title("Growth of an E. coli culture")
````

## Points are joined in list order

`plt.plot()` joins the points in the order of the lists. It does not sort them
by x. Before you run this cell, sketch on paper what you expect: the first
point is at (1, 0), the second at (2, 4), and so on.

````python
x = [1, 2, 3, 0, 4, 1]
y = [0, 4, 0, 3, 3, 0]
plt.plot(x, y)
````

## Points instead of a line

An optional third argument, the *format string*, sets how the data are drawn.
`"o"` draws a circle at each point and no line. `"o-"` draws circles joined by
a line. Measurements are often best shown as points.

````python
plt.plot(hours, od600, "o")
plt.xlabel("time (h)")
plt.ylabel("OD600")
````

````python
plt.plot(hours, od600, "o-")
plt.xlabel("time (h)")
plt.ylabel("OD600")
````

## Two lines in one plot

Call `plt.plot()` twice in the same cell. Give each line a `label`, and call
`plt.legend()` to show the labels.

````python
od600_mutant = [0.05, 0.06, 0.09, 0.14, 0.22, 0.33, 0.45]

plt.plot(hours, od600, "o-", label="wild type")
plt.plot(hours, od600_mutant, "o-", label="mutant")
plt.xlabel("time (h)")
plt.ylabel("OD600")
plt.legend()
````

`label="wild type"` is a new way to give an argument: by its *name*. You will
learn more about this later. For now, copy the pattern.

## Only y values

If you give `plt.plot()` only one list, it uses that list for y. For x it uses
the indexes 0, 1, 2, …

````python
plt.plot(od600)
````

## When x and y do not match

Each point needs one x and one y, so the two lists must have the same length.
Predict, then run:

````python
plt.plot([0, 1, 2], [5, 6])
````

The last line of the error message says that `x` and `y` must have the same
first dimension: here, `x` has 3 items but `y` has only 2.

## Keep your code tidy

In Jupyter, a plot appears without any extra command. You may see
`plt.show()` in code written for plain Python programs (we meet those in week
5). Watch out for `plt.show` *without* round brackets: it names the function
but does not call it, so it does nothing. Lines that do nothing only make your
code harder to read, so delete them. For the same reason, delete the
`# YOUR CODE HERE` comment in an exercise cell once you have written your
code.

### More format strings

The format string is a short code that can combine three things:

- colour: `"r"` red, `"g"` green, `"b"` blue, `"k"` black
- marker at each point: `"o"` circles, `"*"` stars, `"^"` triangles
- line: `"-"` solid, `"--"` dashed, `":"` dotted. If the format string has no
  line character, no line is drawn.

For example, `"r*:"` draws red stars joined by a dotted line. You do not need
to memorise these; look them up when you need them.

````python
plt.plot(hours, od600, "r*:")
plt.xlabel("time (h)")
plt.ylabel("OD600")
````
