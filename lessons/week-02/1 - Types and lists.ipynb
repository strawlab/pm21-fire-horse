## Types

Every value in Python has a **type**. You rarely see the type written, but it
decides what you can do with a value. Python decides the type from exactly how
the value is written. These three look similar, but they have three different
types:

```python
10
10.0
"10"
```

The function `type()` tells you the type of a value.

````python
type(10)
````

````python
type(10.0)
````

````python
type("10")
````

- An `int` (integer) is a whole number, such as `10` or `-32`. Integers are
  exact.
- A `float` (floating-point number) is a number with a decimal point, such as
  `10.0` or `332.6`. Because of the way computers store them, floats are often
  only very close approximations. We come back to this later.
- A `str` (string) is text: here, the two characters `1` and `0`. A string is
  written in quotes. Single quotes and double quotes both work.

````python
type('10')
````

Notice that in `x = 10` you never write that `x` is an integer. Python works
out the type from exactly which characters you typed. This keeps the code
short, but it also hides something important: the same program can behave
differently depending on whether a value is an `int`, a `float` or a `str`.
Many beginner problems come down to a value having a different type than you
thought. When in doubt, check with `type()`.

### Why types matter

The same operator can do different things for different types. For numbers,
`+` adds. For strings, `+` joins the two strings together. This is called
**concatenation**.

````python
10 + 10
````

````python
"10" + "10"
````

What happens if you mix a string and a number? Predict, then run.

````python
"10" + 10
````

A `TypeError` means that an operation got a value of the wrong type. Python
does not guess whether you meant the number 10 or the text "10". Next week you
will learn how to convert a value from one type to another.

### A variable does not have a fixed type

The *value* has a type. A variable is only a name for a value, and the same
name can later refer to a value of a different type.

````python
x = 10
print(type(x))
x = "ten"
print(type(x))
````

### Strings

A string can contain almost any character, including umlauts and emoji.

````python
word = "Übersetzen 😀"
word
````

A string can also contain quotes. If the text contains double quotes, write
the string with single quotes around it:

````python
note = 'the primer is "GCATCCCGGG"'
print(note)
````

This version gives an error. Before you run it, try to say why.

````python
note = "the primer is "GCATCCCGGG""
print(note)
````

Python thinks the string ends at the second `"`, and then it cannot make sense
of the rest of the line. One fix is the single quotes above. Another fix is to
put a backslash `\` in front of each inner quote. The backslash is an
*escape character*: it tells Python that the next character is part of the
string.

````python
note = "the primer is \"GCATCCCGGG\""
print(note)
````

````python
type(None)
````

`None` is written with a capital N and without quotes. `"None"` with quotes is

### Where `None` comes from

You meet `None` most often as the return value of a function that has no
`return`. Here someone forgot the `return`:

````python
def to_celsius(fahrenheit):
    (fahrenheit - 32) * 5 / 9

temp = to_celsius(98.6)
print(temp)
````

The function computes the right number, but then throws it away, and returns
`None`. Often you only notice later, when you try to use the result. Predict
what happens:

````python
temp + 1
````

Read the error message: `NoneType` and `int`. When you see `NoneType` in an
error message, a value you expected to be something else is `None`. Very often,
the reason is a missing `return`. Add `return` in front of the calculation in
`to_celsius`, run both cells again, and check that the error is gone.

If the result of a cell is `None`, Jupyter shows nothing. Use `print` to see

## Lists

A **list** holds several values in a fixed order. You write a list with square
brackets, with commas between the items.

````python
temperatures = [21.5, 22.0, 23.1, 22.8]
temperatures
````

````python
type(temperatures)
````

A list can contain values of different types, and even other lists. A list

````python
mixed = [1, "two", 3.0, None, [4, 5]]
mixed
````

````python
empty = []
empty
````

The built-in function `len()` returns the number of items in a list (its

````python
len(temperatures)
````

````python
len(empty)
````

### Adding to a list with `append`

`.append()` adds one item to the end of a list. `append` is a **method**: a
function that belongs to a value. You call it by writing a dot after the value
(or the variable name), then the method name.

`append` changes the list itself. Watch the list grow:

````python
bases = []
bases.append("A")
print(bases)
bases.append("C")
print(bases)
bases.append("G")
print(bases)
````

What does `append` *return*? Predict, then run.

````python
result = bases.append("T")
print(bases)
print(result)
````

`append` changes the list, and returns `None`. This leads to a common bug:

```python
bases = bases.append("T")
```

After this line, `bases` is `None` and your list is gone. Just write
`bases.append("T")`.

## Indexing: getting one item

To get one item from a list, write its position in square brackets right
after the list. This is called *get item*. The position is called the
**index**. Python counts from **0**, so the first item has index 0. Some other
programming languages count from 1. Neither choice is better, but you must
know which one your language uses.

````python
colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
colors[0]
````

````python
colors[2]
````

A negative index counts from the end: `-1` is the last item, `-2` the one
before it, and so on.

````python
colors[-1]
````

````python
colors[-3]
````

The index can come from a variable:

````python
i = 4
colors[i]
````

The list `colors` has 7 items. Predict what happens here:

````python
colors[7]
````

The items have the indexes 0 to 6, so index 7 does not exist: `IndexError`.

The index must be an integer. Predict, then run:

````python
colors["blue"]
````

### Changing an item

You can also put an index on the *left* side of `=`. Then it is not *get
item* but *set item*: it replaces that item.

````python
colors[3] = "GREEN"
colors
````

Lists can be changed after they are created. We say lists are **mutable**.

## Tuples

A **tuple** is like a list, but it is written with round brackets, and it
cannot be changed after it is created: tuples are **immutable**. Tuples are
often used for a few values that belong together, such as the x and y of a
point. You will see them often in other people's code.

````python
point = (3, 4)
print(type(point))
point
````

Indexing works as for lists:

````python
point[0]
````

But changing an item does not. Predict, then run:

````python
point[0] = 10
````

One trap: a tuple with a single item needs a comma. Without the comma, the
round brackets are just brackets, as in maths.

````python
a = (5)
b = (5,)
print(type(a))
print(type(b))
````

Why does Python have both lists and tuples? Because lists can be changed and
tuples cannot. The next section shows that changeable values can surprise
you. Values that cannot be changed avoid these surprises.

## Two names, one list

For numbers, assignment works as you would expect. Predict both values:

````python
x = 10
y = x
y = 3
print(x)
print(y)
````

Now the same idea with a list. Predict both lines before you run the cell.

````python
x = ["A", "C", "G"]
y = x
y[0] = "T"
print(y)
print(x)
````

Were you surprised? `y = x` does **not** copy the list. It gives the *same*
list a second name. If you change the list through one name, you see the
change through the other name too.

Paste both cells above into the
[Python Tutor](https://pythontutor.com/visualize.html) and step through them.
The Tutor draws each name with an arrow to its value:

- In the first cell, `y = 3` points `y` at a new value. The value 10 is not
  changed, so `x` is still 10.
- In the second cell, `y[0] = "T"` changes the list itself. Both arrows point
  at that one list.

If you want an independent copy, use the method `.copy()`. If you are not
sure whether you need a copy, make one.

````python
x = ["A", "C", "G"]
y = x.copy()
y[0] = "T"
print(y)
print(x)
````

### Functions can change lists

The same thing happens when you pass a list to a function: the argument is
another name for the same list. A function can change the list, even if it
returns nothing. This is a *side effect*.

````python
def add_stop_codon(codons):
    codons.append("TAA")

my_codons = ["ATG", "GCT", "GGA"]
result = add_stop_codon(my_codons)
print(my_codons)
print(result)
````

### Operators are function calls in disguise

`+` is an **operator**. More exactly, it is a *binary* operator: it takes two
values, one on the left and one on the right. Behind the scenes, Python looks
at the type of the left value and calls a special function that belongs to
that type. You never need to write it this way yourself, but it shows why the
type decides what `+` does:

````python
int.__add__(3, 4)
````

````python
str.__add__("abc", "def")
````

### Several steps in one line

Programs get complicated quickly when one line does several things. What is
the value of `y`? Predict, then run.

````python
y = 4 * 3 + 2
y
````

As in school maths, `*` comes before `+`, so the result is 14, not 20. Use
round brackets to change the order: `4 * (3 + 2)` is 20.

To work out `4 * 3 + 2`, Python first computes `4 * 3` and keeps the result 12
somewhere, then adds 2. It is almost as if you had written two lines:

````python
tmp = 4 * 3
y = tmp + 2
y
````

There is one difference: the two-line version creates a real variable `tmp`.
If you already had a variable called `tmp` holding important data, the
two-line version overwrites it, but the one-line version does not. When you
read a line of code, try to see each step it contains.

### Why can a function change a list, but not a number?

Above, `add_stop_codon` changed the list it was given. Now compare a function
that gets a number and assigns a new value to its argument. Predict both
printed values:

````python
def add_four(a):
    a = a + 4

x = 10
result = add_four(x)
print(x)
print(result)
````

`x` is still 10. Inside the function, `a = a + 4` points the local name `a` at
a new value, 14. That does not change the value 10, so `x` is not affected.
(You saw this in week 1.) The function has no `return`, so `result` is
`None`.

The rule: *assigning* to a name (`a = ...`) never changes a value that other
names point to. *Changing* a list (with `.append()`, or with
`codons[0] = ...`) does, and every name for that list sees the change. Step
through both functions in the Python Tutor to see the difference.
