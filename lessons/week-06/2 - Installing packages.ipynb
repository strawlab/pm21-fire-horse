## Packages

Python comes with many tools built in. But `matplotlib`, which you have used
for plots since week 2, is not one of them. It is a **package**: extra Python
code, written by other people, that must be installed before `import` can
find it. Most of the Python that scientists use is in packages: `numpy`
(maths with many numbers), `pandas` (data tables), `matplotlib` (plots), and
many thousands more.

On your own computer, Sunbeam installed the packages for you. On the course
server, we installed them. On a new computer, nobody has done this. firehorse
is such a computer, and today you install a package yourself.

## A script that needs a package

Go to your directory for this week and write a small script with `nano`

## Environments

You cannot install packages into the Python of firehorse itself. It belongs
to the whole computer, and normal users are not allowed to change it. That is
a good thing: on a shared computer, one person's packages should not break
another person's work.

Instead, you make your own **environment**: a directory in your home that
holds its own Python together with the packages *you* install. You can have
several environments, for example one per project, each with its own
packages.

The tool we use to make environments and to install packages is called
**`uv`**. It is not on firehorse yet either. So there are three steps:

1. Install `uv` (once).
2. Make an environment (once).
3. Install packages into the environment (whenever you need a new one).

## Step 1 (once): install `uv`

Type this line exactly. It is easiest to copy and paste it:

```
$ curl -LsSf https://astral.sh/uv/install.sh | sh
```

You should see something like this (the version number may be different):

```
downloading uv 0.13.9 x86_64-unknown-linux-gnu
installing to /home/yourname/.local/bin
  uv
  uvx
everything's installed!

To add $HOME/.local/bin to your PATH, either restart your shell or run:

    source $HOME/.local/bin/env (sh, bash, zsh)
    source $HOME/.local/bin/env.fish (fish)
```

What did this line do?

- `curl` downloads something from the internet. Here it downloads the file
  `install.sh` from `astral.sh`, the website of the people who make `uv`.
- `|` (the "pipe") gives what `curl` downloaded to the next program.
- `sh` runs it as a list of commands.

So this line **downloads a program from the internet and runs it at once**.
That is convenient, and it is also dangerous: the program can do everything
that you can do on this computer. Only run such a line if you trust the
website it comes from. Never run one that you do not understand, from a
forum or a chat. We do it here, once, because this is the official way to
install `uv`.

The program put `uv` into your home directory, in `/home/yourname/.local/bin`.
The terminal that is open now does not look there yet. Do what the message
says:

```
$ source $HOME/.local/bin/env
```

(Logging out and in again has the same effect. Terminals that you open
later find `uv` by themselves.) Check that it worked:

```
$ uv --version
uv 0.13.9 (x86_64-unknown-linux-gnu)
```

## Step 2 (once): make an environment

```
$ uv venv ~/course
Using CPython 3.13.5 interpreter at: /usr/bin/python3
Creating virtual environment at: /home/yourname/course
Activate with: source /home/yourname/course/bin/activate
```

`uv venv` makes a new environment ("virtual environment", short *venv*) in
the directory `~/course`. Look into it with `ls ~/course` if you like. Do not
change anything in there by hand.

## Activate the environment (every time you log in)

An environment does nothing until you switch it on. This is called
**activating** it:

```
$ source ~/course/bin/activate
```

Your prompt changes. It now starts with the name of the environment:

```
(course) yourname@firehorse:~/week-06$
```

`(course)` means: in this terminal, `python3` is now the Python of your
environment. `which` shows which program a command name stands for:

```
(course) $ which python3
/home/yourname/course/bin/python3
```

Without the environment it would print `/usr/bin/python3`, the Python of the
whole computer.

**You must activate the environment again every time you log in**, and in
every new terminal. Look for `(course)` in the prompt. It is the first thing
to check when Python cannot find a package.

## Step 3: install a package

With the environment active:

```
(course) $ uv pip install numpy
Using Python 3.13.5 environment at: /home/yourname/course
Resolved 1 package in 147ms
Downloading numpy (15.0MiB)
 Downloaded numpy
Prepared 1 package in 181ms
Installed 1 package in 14ms
 + numpy==2.5.5
```

`uv` downloaded `numpy` from the internet and put it into your environment.
The download comes from the *Python Package Index* (https://pypi.org), where
almost all Python packages are published. The last line shows the package and
its version.

Now run the script again:

```
(course) $ python3 mean.py
mean wing length: 2.625
```

Some packages need other packages. `uv` then installs those as well, and the
list at the end is longer. You will see this in the exercise.

## What happens without the environment

Switch the environment off with `deactivate`, and try again:

```
(course) $ deactivate
$ python3 mean.py
Traceback (most recent call last):
  File "/home/yourname/week-06/mean.py", line 1, in <module>
    import numpy as np
ModuleNotFoundError: No module named 'numpy'
```

The same error as at the start. `numpy` is still installed, but only in your
environment, and this `python3` is the one of the whole computer. Activate
the environment again and it works:

```
$ source ~/course/bin/activate
(course) $ python3 mean.py
mean wing length: 2.625
```

## When something goes wrong

- `uv: command not found`: the terminal does not know where `uv` is. Run
  `source $HOME/.local/bin/env`, or log out and in again.
- `ModuleNotFoundError` although you installed the package: the environment
  is not active. Is `(course)` in your prompt? If not, run
  `source ~/course/bin/activate`.
- `No such file or directory` for `~/course/bin/activate`: you have not made
  the environment yet (step 2), or you gave it another name.
- `uv pip install` cannot find a package: check the spelling of its name.
  The name to install is usually the name you `import`, but not always.

## Summary

| When | Command | What it does |
| ---- | ------- | ------------ |
| once | `curl -LsSf https://astral.sh/uv/install.sh \| sh` | install `uv` |
| once | `uv venv ~/course` | make the environment `~/course` |
| every login | `source ~/course/bin/activate` | switch the environment on |
| when needed | `uv pip install NAME` | install the package `NAME` into the active environment |
| | `which python3` | show which Python you are using |
| | `deactivate` | switch the environment off |

- A package must be installed before `import` can find it.
- An environment is a directory with its own Python and its own packages.
- `(course)` in the prompt shows that the environment is active.

## Going further (optional)

- `uv pip list` lists the packages in the active environment, with their
  versions.
- A version can be chosen: `uv pip install "numpy==2.3.0"`. Scientists write
  down the versions of the packages they used, so that an analysis can be
  repeated later with the same versions.
- How does the terminal find programs like `ls`, `python3` or `uv`? It looks
  through a list of directories, the **`PATH`**. `echo $PATH` prints it.
  Activating an environment puts `~/course/bin` at the front of this list.
  That is all "activating" does, and it is why `which python3` changes.
- You may see `pip install` or `conda install` in other tutorials. They are
  older tools for the same job. The ideas (packages, environments) are the
  same.
