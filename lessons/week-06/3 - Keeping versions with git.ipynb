## Why keep versions?

You know file names like `thesis_final.docx`, `thesis_final2.docx` and
`thesis_really_final.docx`. With code it gets worse. A script worked
yesterday. Today you changed three things, and now it gives wrong numbers.
What exactly did you change? Can you go back?

**git** is a program that keeps the versions for you. With git you can

- save a version of your files whenever you like, with a short note on what
  you changed,
- see exactly what changed between two versions,
- go back to a saved version when you have broken something,
- get code from other people, together with its whole history.

Almost all scientific software is developed with git. And next week, an AI
assistant will change your files for you. Then the habit from this lesson
becomes important: **save a version before you let anybody else change your
code.** If the change is bad, you go back.

## Three words

- A **repository** (short: *repo*) is a directory whose versions git keeps.
- A **commit** is one saved version: a snapshot of the files, with a
  message, an author and a date.
- **To commit** is to save such a version.

## Make a repository: `git init`

Make a new directory in `~/week-06` and turn it into a repository:

```
$ cd ~/week-06
$ mkdir dna-tools
$ cd dna-tools
$ git init
Initialized empty Git repository in /home/yourname/week-06/dna-tools/.git/
```

`git init` made a hidden directory `.git` (see it with `ls -a`). git keeps
all versions in there. Never change or delete anything in `.git` by hand.

The most useful git command is `git status`. It tells you what git sees, and
it changes nothing. Use it all the time:

```
$ git status
On branch main

No commits yet

nothing to commit (create/copy files and use "git add" to track)
```

## The first commit: `git add` and `git commit`

Write a small script with `nano` (Ctrl-O and Enter to save, Ctrl-X to exit):

```
$ nano count.py
```

```python
seq = "GATTACA"
print("length:", len(seq))
```

Run it to check that it works:

```
$ python3 count.py
length: 7
```

Now ask git what it sees:

```
$ git status
On branch main

No commits yet

Untracked files:
  (use "git add <file>..." to include in what will be committed)
	count.py

nothing added to commit but untracked files present (use "git add" to track)
```

*Untracked* means: there is a file here, but git does not keep versions of
it yet. Saving a version takes **two steps**:

1. `git add` chooses which files go into the next commit.
2. `git commit` saves the commit, with a message.

Think of a group photo. `git add` puts people into the picture. `git commit`
presses the button.

```
$ git add count.py
$ git status
On branch main

No commits yet

Changes to be committed:
  (use "git rm --cached <file>..." to unstage)
	new file:   count.py
```

```
$ git commit -m "First version: print the length"
[main (root-commit) bd275f0] First version: print the length
 1 file changed, 2 insertions(+)
 create mode 100644 count.py
```

`-m` is followed by the **message**, in quotes: a short note on what this
version does or what you changed. Write it for yourself in six months.

`bd275f0` is the **id** of the commit. Every commit has its own id. Yours is
different from the one printed here.

```
$ git status
On branch main
nothing to commit, working tree clean
```

*Clean* means: the files are the same as in the last commit.

## Change, look, commit

Open the script again with `nano count.py` and add a third line:

```python
print("number of A:", seq.count("A"))
```

Run it (`python3 count.py`). Then ask git:

```
$ git status
On branch main
Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	modified:   count.py

no changes added to commit (use "git add" and/or "git commit -a")
```

git has noticed that `count.py` is *modified*. `git diff` shows **what**
changed since the last commit:

```
$ git diff
diff --git a/count.py b/count.py
index e364a4f..ea32f47 100644
--- a/count.py
+++ b/count.py
@@ -1,2 +1,3 @@
 seq = "GATTACA"
 print("length:", len(seq))
+print("number of A:", seq.count("A"))
```

Skip the first lines. Read the lines at the bottom:

- a line starting with `+` was **added**,
- a line starting with `-` was **removed**,
- the other lines are unchanged. They are shown so that you see where you
  are in the file.

A changed line shows up as one `-` line (the old text) and one `+` line (the
new text).

If the change is what you wanted, commit it. The same two steps:

```
$ git add count.py
$ git commit -m "Also count the A bases"
[main 1610ee7] Also count the A bases
 1 file changed, 1 insertion(+)
```

This is the rhythm of working with git: **change, check that it works,
`git diff`, `git add`, `git commit`**. Commit often. A commit costs nothing.

## Looking at the history: `git log`

```
$ git log
commit 1610ee74a11e82c31087e8110ffdea617242b3e2
Author: Anna Beispiel <anna.beispiel@students.uni-freiburg.de>
Date:   Fri Nov 20 09:02:58 2026 +0100

    Also count the A bases

commit bd275f08c779393c37f63eac13e185e1cbbcf633
Author: Anna Beispiel <anna.beispiel@students.uni-freiburg.de>
Date:   Fri Nov 20 09:02:58 2026 +0100

    First version: print the length
```

The newest commit is at the top. Each commit has its id (the long one; the
first seven characters are enough to name it), the author, the date and the
message.

If the history is longer than the window, git shows it page by page, with
`less` from last week. Press `q` to quit.

A short form, one line per commit:

```
$ git log --oneline
1610ee7 Also count the A bases
bd275f0 First version: print the length
```

`git show` with an id shows one commit together with what it changed. Use
one of **your** ids:

```
$ git show 1610ee7
```

## Going back: `git restore`

Now break the script on purpose. Open it with `nano`, delete the second line
and add a line with a mistake, so that the file looks like this:

```python
seq = "GATTACA"
print("number of A:", seq.count("A"))
print("oops"
```

```
$ python3 count.py
  File "/home/yourname/week-06/dna-tools/count.py", line 3
    print("oops"
         ^
SyntaxError: '(' was never closed
```

`git diff` shows what you did since the last commit:

```
$ git diff
diff --git a/count.py b/count.py
index ea32f47..c0651b4 100644
--- a/count.py
+++ b/count.py
@@ -1,3 +1,3 @@
 seq = "GATTACA"
-print("length:", len(seq))
 print("number of A:", seq.count("A"))
+print("oops"
```

You could repair this by hand. But there is a faster way. `git restore`
brings a file back to how it was in the last commit:

```
$ git restore count.py
$ cat count.py
seq = "GATTACA"
print("length:", len(seq))
print("number of A:", seq.count("A"))
```

This also works for a file you have deleted:

```
$ rm count.py
$ ls
$ git status
On branch main
Changes not staged for commit:
  (use "git add/rm <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	deleted:    count.py

no changes added to commit (use "git add" and/or "git commit -a")
$ git restore count.py
$ ls
count.py
```

Last week you learned that `rm` has no wastebasket. With git, a file that
you have committed can be brought back.

**Be careful:** `git restore` throws away everything you changed in that
file since the last commit, and those changes cannot be brought back. Look
at `git diff` first. And git can only bring back what you have committed:
one more reason to commit often.

## When something goes wrong

- `fatal: not a git repository`: you are not in a repository. Check with
  `pwd`, and `cd` into the right directory.
- `Author identity unknown` and `Please tell me who you are`: you have not
  told git your name yet. See "Tell git who you are" above, then commit
  again.
- `nothing to commit, working tree clean`: nothing has changed since the
  last commit. Did you save the file in nano?
- `no changes added to commit`: you changed a file but forgot `git add`.
- After `git commit` without `-m`, an editor (nano) opens and waits for a
  message. Type a message in the first line, then Ctrl-O, Enter, Ctrl-X.
- The terminal shows `:` or `(END)` at the bottom and does not react: you
  are in `less`, after `git log` or `git diff`. Press `q`.

## Summary

| Command | What it does |
| ------- | ------------ |
| `git init` | turn the working directory into a repository |
| `git status` | show what git sees (changes nothing) |
| `git add FILE` | choose `FILE` for the next commit |
| `git commit -m "message"` | save a version |
| `git diff` | show what changed since the last commit |
| `git log`, `git log --oneline` | show the history |
| `git show ID` | show one commit and what it changed |
| `git restore FILE` | bring `FILE` back to the last commit |
| `git clone URL` | copy a repository from the internet |
| `git pull` | fetch new commits into a clone |

The rhythm: change, check that it works, `git diff`, `git add`,
`git commit`.

## Going further (optional)

- `git add .` chooses all changed files in the working directory at once.
  Look at `git status` first, so that you know what goes in.
- A file named `.gitignore` lists files that git should not keep, for
  example large data files or plots that a script can make again.
- `git diff ID1 ID2` shows what changed between two commits.
- git is not a backup. Your commits are in the directory `.git`. If the
  directory or the computer is lost, they are lost. People send their
  commits to another computer with `git push`. We do not do this in this
  course.
- git has *branches*: several lines of history side by side. `main` is the
  name of the branch you are on. Branches are how many people work on the
  same code at the same time.
- GitHub is a website that stores git repositories and adds tools for
  working together. git itself runs on your computer and does not need
  GitHub.

**One rule: do not work inside the clone.** Copy what you need into your own
folder, and work on the copy. `cp -r` copies a directory with everything in
it:

```
$ cd ~/PM-21
$ cp -r pm21-fire-horse/exercises/release/exercise-07 .
```

Why? When you open a notebook and save it, the file changes. Sometimes we
correct a notebook after it was published. If you have changed the same file
in your clone, git does not know which version to keep, and `git pull` stops:

```
error: Your local changes to the following files would be overwritten by merge:
	lessons/week-06/1 - Another computer.ipynb
Please commit your changes or stash them before you merge.
Aborting
```

If this happens: `git status` shows which files you changed. If you want to
keep your version of a file, copy it to another folder first. Then bring the
file back to the last commit with `git restore` (put the name in quotes if
it contains spaces, or use Tab), and run `git pull` again:

```
$ git restore "lessons/week-06/1 - Another computer.ipynb"
$ git pull
```

If you never change files inside the clone, this cannot happen.
