Last week you used an index to get **one** item from a list. Often you want

image that contains one cell. Getting a range of items is called **slicing**.

## A slice: start and end

Write a slice with a colon inside the square brackets: `x[start:end]`.

````python
colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
colors[0:3]
````

The slice starts at index 0 and goes **up to, but not including,** index 3.
So you get the items with the indexes 0, 1 and 2, but not `"green"`:

````python
colors[3]
````

A useful side effect of this rule: the number of items in a slice is
`end - start`. `colors[0:3]` has 3 items, and `colors[2:5]` also has 3 items.
Predict, then run:

````python
colors[2:5]
````

Many languages use this rule, but not all. Even experienced programmers forget
such details. When you are not sure, do what they do: test it in a cell.

## Leaving out the start or the end

If you leave out the start, the slice starts at the beginning. If you leave
out the end, the slice goes to the end of the list.

````python
colors[:3]
````

````python
colors[3:]
````

Negative indexes work as well. Remember that `-1` is the last item. Predict
both:

````python
colors[-3:]
````

````python
colors[3:-1]
````

`colors[-3:]` gives the last three items. `colors[3:-1]` goes from index 3 up
to, but not including, the last item.

## Slices are forgiving

An index that is too large gives an `IndexError`, as you saw last week. A
slice that goes past the end does not. It simply stops at the end:

````python
colors[5:100]
````

## Slicing tuples and strings

Slicing works the same for tuples, and for strings. A DNA sequence is often
stored as a string. A slice of a string is again a string.

````python
point = (3, 4, 5)
point[:2]
````

````python
dna = "ATGGCTTAA"
dna[0:3]
````

The first codon is `dna[0:3]`. Predict the second and the third codon:

````python
dna[3:6]
````

````python
dna[6:9]
````

### The defaults are `None`

When you leave out the start or the end, Python actually uses the value
`None` there. These two are the same:

````python
print(colors[:3])
print(colors[None:3])
````

### A step

A slice can have a third number, the *step*. `x[start:end:step]` takes every
step-th item.

````python
colors[0:7:2]
````

A step of `-1` goes backwards. With both start and end left out, `x[::-1]`
reverses a sequence. For DNA, this is the first step towards the reverse
complement.

````python
dna[::-1]
````

### A slice is a new list

A slice of a list is a new list. Changing the slice does not change the
original. So `colors[:]`, a slice of everything, is another way to copy a
list.

````python
first_three = colors[:3]
first_three[0] = "RED"
print(first_three)
print(colors)
````
