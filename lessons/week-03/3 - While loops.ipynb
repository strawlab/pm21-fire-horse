## Repeat while a condition is true

A `while` loop looks like an `if` statement. The difference: after the block
has run, Python goes back and checks the condition again. If it is still
`True`, the block runs again. This repeats until the condition is `False`.

Predict the **last** number this cell prints. Then run it.

````python
y = 0
while y < 1000:
    print(y)
    y = y + 100
print("done")
````

It prints 900, not 1000. When `y` is 900, it is printed, then it becomes 1000.
Then the condition `y < 1000` is `False`, so the loop ends and Python goes on
with the first line after the block.

## A shorter way to update a variable: `+=`

Updating a variable is so common that Python has a short form. `y += 100`
means exactly the same as `y = y + 100`. There are also `-=` and `*=`.

````python
y = 0
while y < 1000:
    print(y)
    y += 100
print("done")
````

````python
z = 10
z *= 3
z -= 5
z
````

## Counting: how long does it take?

A common use of a `while` loop is to count how many steps something takes. A
tree is 1.5 m tall and grows 0.1 m every year. How many years until it is
20 m tall?

````python
height = 1.5
years = 0
while height < 20:
    height += 0.1
    years += 1

print(years)
````

## Building a list in a loop

Often you want to keep every step, not only the final count. Start with
empty lists, and append to them inside the loop. Then you can plot the
result, as you learned last week.

````python
import matplotlib.pyplot as plt

height = 1.5
year = 0
all_years = []
all_heights = []
while height < 20:
    all_years.append(year)
    all_heights.append(height)
    height += 0.1
    year += 1

plt.plot(all_years, all_heights)
plt.xlabel("year")
plt.ylabel("tree height (m)")
````

The tree grows by the same amount every year, so the plot is a straight line:
**linear** growth.

Now suppose the tree grows by 10 % of its height every year instead. Then the
update is `height = height * 1.1`, or `height *= 1.1`. Predict: will it reach
20 m faster or slower? What will the plot look like?

````python
height = 1.5
year = 0
all_years = []
all_heights = []
while height < 20:
    all_years.append(year)
    all_heights.append(height)
    height *= 1.1
    year += 1

plt.plot(all_years, all_heights, "o-")
plt.xlabel("year")
plt.ylabel("tree height (m)")
````

This is **exponential** growth: the bigger the tree, the more it grows.
(Real trees do not grow like this for long. But bacteria in fresh medium, or
an epidemic at its start, do. You will simulate an epidemic in the exercise.)

## Infinite loops

If the condition never becomes `False`, the loop never ends. Here the line
that updates `y` was forgotten:

```python
y = 0
while y < 1000:
    print(y)
```

`y` stays 0 forever, so the cell runs forever. You can see that a cell is
still running by the `[*]` to its left. To stop it, click the square
**stop** button in the toolbar (or use Kernel → Interrupt Kernel). If you
like, copy the code above into a new cell, run it, and stop it.

If one of your cells seems to hang, check: does the loop change something, so
that the condition will eventually become `False`?

## `break`: leave the loop now

`break` leaves the loop immediately. A common pattern is `while True:`, which
on its own would run forever, together with a `break`:

````python
y = 0
while True:
    print(y)
    y += 100
    if y >= 1000:
        break
print("done")
````

This does exactly the same as the first loop in this notebook. Often there
are several ways to write the same thing. Choose the one that is easiest to
read: here, `while y < 1000:`.

### `continue`: skip the rest of this round

`continue` does not leave the loop. It skips the rest of the block and goes
back to the condition check. Predict which numbers this prints:

````python
n = 0
while n < 10:
    n += 1
    if n == 3 or n == 7:
        continue
    print(n)
````

### A loop controlled by a variable

You will also see loops that use a `bool` variable as the condition, and set
it to `False` to stop. This does the same as the first loop of this notebook
again. Step through it in the Python Tutor:

````python
keep_going = True
y = 0
while keep_going:
    print(y)
    y += 100
    if y >= 1000:
        keep_going = False
print("done")
````
