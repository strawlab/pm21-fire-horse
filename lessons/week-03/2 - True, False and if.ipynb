## `if`: run code only when a condition is true

You saw `if` in the Blockly bird game in week 1. In Python, `if` is followed
by a condition and a colon. The indented lines below it form a **block**, just
like the body of a function. The block runs only if the condition is `True`.

````python
ph = 5.5
if ph < 7:
    print("The sample is acidic.")
print("Done.")
````

Change `ph` to `8` and run the cell again. The indented line is skipped, but
`print("Done.")` is not indented, so it always runs.

## `else`: otherwise

An `if` can have an `else` block, which runs when the condition is `False`.

````python
ph = 8
if ph < 7:
    print("acidic")
else:
    print("not acidic")
````

## `elif`: more than two cases

`elif` means "else, if". Python checks the conditions from top to bottom and
runs the **first** block whose condition is `True`. The `else` block runs only
if none of them is `True`.

````python
ph = 7
if ph < 7:
    print("acidic")
elif ph == 7:
    print("neutral")
else:
    print("basic")
````

Try other values of `ph`. Then paste the cell into the
[Python Tutor](https://pythontutor.com/visualize.html) and watch which lines
run.

An `if` statement often sits inside a function:

````python
def classify_ph(ph):
    if ph < 7:
        return "acidic"
    elif ph == 7:
        return "neutral"
    else:
        return "basic"

print(classify_ph(2.0))
print(classify_ph(7))
print(classify_ph(12.5))
````

`return` ends the function at once, so only one of the three `return` lines
runs.

## Combining conditions: `and`, `or`, `not`

- `a and b` is `True` only if both `a` and `b` are `True`.
- `a or b` is `True` if at least one of them is `True`.
- `not a` turns `True` into `False` and `False` into `True`.

For example, is a temperature in the range for growing *E. coli* well?

````python
temperature = 37
if temperature >= 30 and temperature <= 40:
    print("good growth temperature")
else:
    print("too cold or too hot")
````

````python
base = "N"
if not (base == "A" or base == "C" or base == "G" or base == "T"):
    print(base, "is not one of A, C, G, T")
````

## Checking assumptions with `assert`

`assert` checks that a condition is `True`. If it is, nothing happens. If it
is `False`, Python stops with an `AssertionError`. Every check cell in the
exercises works this way.

````python
assert 2 + 2 == 4
````

````python
assert 1 + 1 == 3, "my arithmetic is wrong"
````

You can use `assert` in your own functions, to state what the function
expects. A mistake is then caught early, with a clear message, instead of
causing a strange result later:

````python
def classify_ph_checked(ph):
    assert ph >= 0 and ph <= 14, "pH must be between 0 and 14"
    return classify_ph(ph)

classify_ph_checked(70)
````

## Asking the user: `input()`

`input()` shows a message, waits until the user types something and presses
Enter, and returns what was typed. Run the cell below, type a number into the
box that appears, and press Enter.

````python
answer = input("How many base pairs do you want to sequence? ")
print(type(answer))
````

`input()` always returns a **string**, even if you typed digits. To compute

## Other values as conditions

An `if` condition does not have to be a `bool`. Python converts the value
to `True` or `False` for you. The rules: `0`, `0.0`, the empty string `""`,
the empty list `[]` and `None` count as `False`. Almost everything else counts
as `True`. You can check with `bool()`:

````python
print(bool(0))
print(bool(42))
print(bool(""))
print(bool("hello"))
print(bool([]))
````

````python
samples = []
if samples:
    print("we have samples")
else:
    print("no samples yet")
````

### Surprising conversions

Only the empty string converts to `False`. Every other string converts to
`True`, even the string `"False"`. Predict each line, then run:

````python
print(bool(" "))
print(bool("False"))
print(bool(str(False)))
````

`True` and `False` also behave like the numbers 1 and 0. This goes back to the
early days of computers, where one bit of memory could mean either "true or
false" or "1 or 0".

````python
print(int(True))
print(int(False))
print(2 + False)
````

Several rules can act in one line. Predict both:

````python
print(2 + 2 == 0)
print(2 + (2 == 0))
````

In the first line, `2 + 2` is computed first, then compared with 0: `False`.
In the second line, the brackets make Python compare `2 == 0` first, which
gives `False`, and `2 + False` is 2. Other programming languages would refuse
to add a number and a `bool`.

### Nested blocks

A block can contain another `if` with its own block. The indentation shows
which line belongs to which block. Predict what is printed, then step through
it in the Python Tutor:

````python
a = 1
b = -2

if a == 1:
    if b > 0:
        print("a is one and b is positive")
    else:
        print("a is one and b is not positive")
else:
    print("a is not one")
````
