A list stores items by *position*: item 0, item 1, … Often you want to look
things up by a *name* instead: the amino acid for a codon, the measurement for
a sample, the count for each base. For this, Python has the **dictionary**
(type `dict`). Like a paper dictionary (*Wörterbuch*), it stores **keys**
(the words you look up) and their **values** (what you find).

## Creating a dictionary and looking things up

Write a dictionary with curly braces. Each entry is `key: value`, and entries
are separated by commas.

````python
codon_table = {"ATG": "Met", "TGG": "Trp", "TAA": "Stop"}
codon_table
````

To look up a value, write the key in square brackets, as with a list index:

````python
codon_table["TGG"]
````

Looking up a key that is not in the dictionary gives a `KeyError`. Predict,
then run:

````python
codon_table["GGG"]
````

## Adding and changing entries

With a key on the *left* of `=`, you add a new entry, or change an existing
one:

````python
codon_table["GGG"] = "Gly"
codon_table["TAA"] = "STOP"
codon_table
````

`len()` gives the number of entries. The operator `in` checks whether a key
is in the dictionary. It is often used together with `if`:

````python
print(len(codon_table))
print("ATG" in codon_table)
print("CCC" in codon_table)
````

````python
codon = "CCC"
if codon in codon_table:
    print(codon_table[codon])
else:
    print(f"{codon} is not in our table")
````

## Looping over a dictionary

A `for` loop over a dictionary goes through its **keys**. Use each key to look
up its value:

````python
od600 = {"A1": 0.31, "A2": 0.45, "A3": 0.12}
for sample in od600:
    print(f"{sample}: {od600[sample]}")
````

Values can be anything, also lists. Keys are usually strings or numbers.

````python
replicates = {"control": [0.30, 0.32, 0.29], "treated": [0.12, 0.15, 0.11]}
replicates["treated"]
````

## The counting pattern

A very common job: count how often each thing occurs. Start with an empty
dictionary. For each item: if it is already a key, add 1 to its count;
otherwise, add it with the count 1.

````python
dna = "GATTACAGATTACA"
counts = {}
for base in dna:
    if base in counts:
        counts[base] += 1
    else:
        counts[base] = 1

counts
````

Step through this cell in the Python Tutor. Watch the dictionary grow.

## Building a new dictionary from another one

Loop over one dictionary, compute something for each key, and store it in a
new dictionary under the same key:

````python
replicates = {"control": [0.30, 0.32, 0.29], "treated": [0.12, 0.15, 0.11]}
means = {}
for condition in replicates:
    values = replicates[condition]
    means[condition] = sum(values) / len(values)

means
````

### Keyword arguments

In week 2 you wrote `plt.plot(x, y, label="wild type")`. `label` is a
*keyword argument*: an argument given by name, which has a default value when
you leave it out. You can make your own:

````python
def final_volume(stock_volume, dilution_factor=10):
    return stock_volume * dilution_factor

print(final_volume(5))
print(final_volume(5, dilution_factor=100))
````

Keyword arguments let a function have sensible defaults for the usual case,
and options for special cases.

### Reading a CSV file by hand

Data tables are often stored as **CSV** files ("comma-separated values"):
each line is a row, and the values in a row are separated by commas. The file
`iris.csv` next to this notebook holds measurements of 150 iris flowers. (Make
sure you downloaded `iris.csv` together with this notebook.)

With what you know now, you can read it into a dictionary with one key per
column and a list of values for each column. In week 7 you will learn the
`pandas` package, which does this in one line. Doing it once by hand shows
what such a package does for you.

`open()` opens a file, and `.readlines()` gives a list with one string per
line. Look at the first lines:

````python
lines = open("iris.csv").readlines()
for line in lines[:4]:
    print(repr(line))
````

(`repr()` shows a string as Python would write it, so you can see the `\n`,
the line break, at the end of each line.)

The first line holds the column names. Each of the other lines holds one
flower. We use `.strip()`, `.split()`, `float()` and the dictionary patterns
from above:

````python
lines = open("iris.csv").readlines()
column_names = lines[0].strip().split(",")

iris = {}
for name in column_names:
    iris[name] = []

for line in lines[1:]:
    entries = line.strip().split(",")
    for name, entry in zip(column_names, entries):
        if name != "species":
            entry = float(entry)
        iris[name].append(entry)

print(column_names)
print(iris["sepal_length"][:5])
````

````python
import matplotlib.pyplot as plt

plt.plot(iris["sepal_width"], iris["petal_width"], "o", alpha=0.3)
plt.xlabel("sepal width (cm)")
plt.ylabel("petal width (cm)")
````

`alpha=0.3` makes the points partly transparent, so that points on top of
each other look darker.
