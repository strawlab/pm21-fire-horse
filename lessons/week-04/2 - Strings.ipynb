## Strings are sequences

You already know most of this: a string can be indexed, sliced, measured with
`len()` and looped over, just like a list.

````python
dna = "ATGGAATTCGCTTAA"
print(dna[0])
print(dna[0:3])
print(len(dna))
````

The operator `in` checks whether one string occurs inside another. For
example, does our sequence contain `GAATTC`, the site where the restriction
enzyme EcoRI cuts?

````python
"GAATTC" in dna
````

## String methods

Strings have many methods. You call them with a dot, like `.append()` on a
list. Here are the most useful ones.

`.upper()` and `.lower()` change the case. Useful when sequences come in
mixed case:

````python
seq = "atgGCta"
seq.upper()
````

`.count()` counts how often a piece of text occurs:

````python
"GATTACA".count("A")
````

`.replace()` replaces every occurrence of one piece of text by another. For
example, to transcribe DNA into RNA, replace each `T` by `U`:

````python
dna = "ATGGCTTAA"
rna = dna.replace("T", "U")
rna
````

`.startswith()` and `.endswith()` return `True` or `False`:

````python
print(dna.startswith("ATG"))
print("data_2026.csv".endswith(".csv"))
````

`.strip()` removes spaces, tabs and line breaks from both ends of a string.
Text that was typed or pasted, or read from a file, often has them:

````python
raw = "   ATGGCTTAA \n"
print(len(raw))
clean = raw.strip()
print(len(clean))
clean
````

`.split()` cuts a string into a list of strings, wherever a separator occurs.
A line from a data file often looks like this:

````python
line = "sample_7,0.45,37.0"
parts = line.split(",")
parts
````

Note that the parts are all strings. To compute with them, convert them, as
you learned last week:

````python
od = float(parts[1])
od * 2
````

### Methods return a new string

A string can never be changed (strings are *immutable*, like tuples). So
string methods do not change the string. They return a **new** string. If you
want to keep the result, assign it to a variable. Predict what this prints:

````python
seq = "atgc"
seq.upper()
print(seq)
````

````python
seq = "atgc"
seq = seq.upper()
print(seq)
````

## f-strings: putting values into text

Very often you want to put values into a piece of text: a message, a plot
title, a line in a report. Write an `f` directly before the opening quote.
Inside the string, anything in curly braces `{ }` is replaced by its value:

````python
name = "nest 1"
n_eggs = 94
message = f"In {name} we found {n_eggs} eggs."
print(message)
````

Inside the braces you can write any expression, for example a calculation:

````python
n_male = 41
print(f"Fraction male: {n_male / n_eggs}")
````

That is too many digits. After a colon, you can give a *format*: `.2f`
means "a number with 2 digits after the decimal point".

````python
print(f"Fraction male: {n_male / n_eggs:.2f}")
````

f-strings are also handy for plot labels and titles, for example
`plt.title(f"Growth at {temperature} °C")`.

### The reverse complement

The other strand of a DNA double helix is the *reverse complement*: each base
is replaced by its partner (A↔T, C↔G), and the order is reversed. Here is one
way, with a loop, `if`/`elif` and `+`. `[::-1]` reverses a string (see the
slicing lesson from week 3):

````python
def reverse_complement(dna):
    result = ""
    for base in dna:
        if base == "A":
            result += "T"
        elif base == "T":
            result += "A"
        elif base == "C":
            result += "G"
        else:
            result += "C"
    return result[::-1]

reverse_complement("ATGGCTTAA")
````

### Older ways to format strings

You will see two older styles in existing code. They do the same as
f-strings:

````python
n = 94
print("We found %d eggs." % n)
print("We found {} eggs.".format(n))
print(f"We found {n} eggs.")
````

To write a literal curly brace in an f-string, double it: `f"{{ {n} }}"`
gives `{ 94 }`.
