Last week, a `while` loop repeated a block *while a condition was true*. Very
often, you want something simpler: do the same thing *once for each item*,
for example for each measurement in a list, or for each base in a DNA
sequence. That is what the `for` loop does. Most loops in Python programs are
`for` loops.

## Looping over a list

The `for` line names a variable, here `t`, and a list. The block runs once for
each item of the list. Each time, `t` is set to the next item.

````python
temperatures = [21.5, 22.0, 23.1, 22.8]
for t in temperatures:
    print(t)
print("done")
````

## Looping over a string

A string is a sequence of characters, so a `for` loop can go through it
character by character:

````python
dna = "GATTACA"
for base in dna:
    print(base)
````

## Counting with `range`

`range(n)` produces the whole numbers 0, 1, 2, … up to, but not including,
`n`. That is exactly `n` numbers. Use it when you want to repeat something a
fixed number of times.

````python
for i in range(5):
    print(i)
````

`range` takes up to three arguments, just like a slice: start, end (not
included) and step. Predict both, then run:

````python
for i in range(2, 6):
    print(i)
````

````python
for year in range(0, 30, 10):
    print(year)
````

## Pattern 1: adding things up

A very common pattern: start a variable at 0 *before* the loop, and add to it
*inside* the loop.

````python
temperatures = [21.5, 22.0, 23.1, 22.8]
total = 0
for t in temperatures:
    total += t

mean = total / len(temperatures)
print(total)
print(mean)
````

Why `89.39999999999999` and not `89.4`? Remember from week 2: computers store
most decimal numbers only approximately, and tiny errors can show up after
additions. For measurements this does not matter. When you show results,
round them, for example with `round(total, 2)`.

## Pattern 2: counting

Combine the loop with an `if` to count only some items. How many bases in this
sequence are G or C?

````python
dna = "GATTACAGGC"
gc_count = 0
for base in dna:
    if base == "G" or base == "C":
        gc_count += 1

print(gc_count)
````

## Pattern 3: building a new list

Start with an empty list before the loop, and append to it inside the loop.
Here we convert temperatures from °C to °F:

````python
celsius = [21.5, 22.0, 23.1, 22.8]
fahrenheit = []
for c in celsius:
    fahrenheit.append(c * 9 / 5 + 32)

print(fahrenheit)
````

These three patterns (add up, count, build a list) cover a large part of
everyday data analysis. Python also has built-in functions for some of
them: `sum(temperatures)` and `max(temperatures)`. Writing the loop yourself
first shows you what such functions do.

## Repeating a simulation step

`range` is also useful to repeat a step a fixed number of times. Last week's
exercise had a function that adds one generation to an epidemic. With a `for`
loop, we can run exactly 5 generations:

````python
def simulate_generation(a, Rt):
    previous = a[-1]
    a.append(previous * Rt)

infected = [1]
for generation in range(5):
    simulate_generation(infected, 2)

print(infected)
````

The loop variable `generation` is not used inside the block. That is fine: we
only need the block to run 5 times.

## `for` or `while`?

- Use `for` when you know what to loop over: the items of a list or string,
  or a fixed number of repetitions.
- Use `while` when you repeat *until something happens*, and you do not know
  in advance how many steps that takes ("until 1000 people are infected").

### Loops inside loops

The items of a list can be lists. A loop inside a loop goes through each of
them:

````python
plates = [[5, 5, 6], [6, 6, 7]]
for plate in plates:
    print("plate:", plate)
    for count in plate:
        print("  colony count:", count)
````

### `break` and `continue` in `for` loops

They work as in `while` loops: `break` leaves the loop, `continue` skips to
the next item. Predict the result:

````python
found = []
for x in range(100):
    if x > 5 and x < 10:
        continue
    if x >= 20:
        break
    found.append(x)
found
````

### The loop variable lives on

After the loop, the loop variable still exists and holds the last item. If you
already had a variable of the same name, it is overwritten. Other languages
keep the loop variable inside the loop; Python does not.

````python
x = "my super important data"
for x in range(3):
    pass
x
````

(`pass` means "do nothing". A block may not be empty, so `pass` fills it.)

### Iterators

Anything a `for` loop can go through is called *iterable*. `range(10)` does
not build a list of 10 numbers in memory: it produces the numbers one at a
time, when the loop asks for the next one. That matters for huge data, such as
a whole genome that does not fit into your computer's memory at once. If you
do want a list, call `list()`:

````python
list(range(10))
````

### `enumerate` and `zip`

`enumerate` gives you the position (index) together with each item. `zip`
walks through two lists side by side. Both give a tuple each time, and the
`for` line can unpack the tuple into two variables at once.

````python
samples = ["A1", "A2", "A3"]
for i, name in enumerate(samples):
    print(i, name)
````

````python
samples = ["A1", "A2", "A3"]
od600 = [0.31, 0.45, 0.12]
for name, od in zip(samples, od600):
    print(name, od)
````
