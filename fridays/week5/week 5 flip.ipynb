```python
bases = "TCAG"     
amino_acids = "FFLLSSSSYY**CC*WLLLLPPPPHHQQRRRRIIIMTTTTNNKKSSRRVVVVAAAADDEEGGGG"

codon_table = {}
i = 0
for b1 in bases:
    for b2 in bases:
        for b3 in bases:
            codon_table[b1 + b2 + b3] = amino_acids[i]
            i += 1
```

```python
def translate(dna_seq):
    """take DNA sequence, as a string, and output amino acid sequence"""
    assert type(dna_seq)==str
    # assert len(dna_seq) % 3 == 0
    i = 0
    peptide_seq = ''
    while True:
        codon = dna_seq[i:i+3]
        i += 3
        if len(codon) != 3:
            break
        aa = codon_table[codon]
        peptide_seq += aa
    return peptide_seq
```

```python
for frame in range(3):
    aa_seq = translate(gfp_dna_seq[frame:])
    print(f"{frame}: {aa_seq}")
```
