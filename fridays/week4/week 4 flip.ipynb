```python
def transcribe(dna_seq):
    """take DNA sequence, as a string, and output mRNA sequence"""
    assert type(dna_seq)==str
    return dna_seq.replace('T','U')

assert transcribe("ATG")=="AUG"
```

```python
bases = "TCAG"     
amino_acids = "FFLLSSSSYY**CC*WLLLLPPPPHHQQRRRRIIIMTTTTNNKKSSRRVVVVAAAADDEEGGGG"

codon_table = {}
i = 0
for b1 in bases:
    for b2 in bases:
        for b3 in bases:
            codon_table[b1 + b2 + b3] = amino_acids[i]
            i += 1
```

```python
def translate(rna_seq):
    """take RNA sequence, as a string, and output amino acid sequence"""
    assert type(rna_seq)==str
    # assert len(rna_seq) % 3 == 0
    i = 0
    peptide_seq = ''
    while True:
        codon = rna_seq[i:i+3]
        i += 3
        if len(codon) != 3:
            break
        aa = codon_table[codon]
        peptide_seq += aa
    return peptide_seq
```

```python
gfp_peptide_seq = translate(gfp_dna_seq)
```

```python
def reverse_complement(dna):
    result = ""
    for base in dna:
        if base == "A":
            result += "T"
        elif base == "T":
            result += "A"
        elif base == "C":
            result += "G"
        else:
            result += "C"
    return result[::-1]

reverse_complement("ATGGCTTAA")
```
