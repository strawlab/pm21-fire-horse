$R_t$ is the average number of people that one infected person infects.
(See [Wikipedia](https://en.wikipedia.org/wiki/Basic_reproduction_number#Effective_reproduction_number).)

large as the one before:

$N_{t+1} = R_t \, N_t$

If $R_t > 1$, the number of infected people grows exponentially, like the
